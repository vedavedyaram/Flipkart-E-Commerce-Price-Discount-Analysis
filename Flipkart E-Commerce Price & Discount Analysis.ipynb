# 3.Checking Outliers for numerical columns

```python
for i in num:
    sns.boxplot(data = data,x = i)
    plt.title(i)
    plt.show()
```

```python
def outlier(df,col_name):

    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.74)

    iqr = q3 - q1

    lb = q1 - 1.5 * iqr
    ub = q3 + 1.5 * iqr

    return df[(df[i] < lb) | (df[i] > ub)]
    
```

```python
for i in num:
    print(i)
    print(len(outlier(data,i)))
```
